```python
import googleapiclient.discovery
import csv
import time
import os
import json

def save_next_page_token(token, filename="next_page_token.json"):
    print(f"Salvataggio nextPageToken: {token}")  # Debug: visualizza il token che si sta salvando
    with open(filename, 'w') as f:
        json.dump({"nextPageToken": token}, f)

def load_next_page_token(filename="next_page_token.json"):
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            data = json.load(f)
            token = data.get("nextPageToken")
            print(f"Caricato nextPageToken: {token}")  # Debug: visualizza il token che viene caricato
            return token
    return None

def search_videos_by_tags(api_key, tags, max_results=500, published_after=None):
    youtube = googleapiclient.discovery.build("youtube", "v3", developerKey=api_key)
    search_query = ' '.join(tags)

    videos = []
    next_page_token = load_next_page_token()

    while len(videos) < max_results:
        try:
            request = youtube.search().list(
                q=search_query,
                part="snippet",
                type="video",
                maxResults=min(max_results - len(videos), 50),
                pageToken=next_page_token,
                publishedAfter=published_after  # Filtro per data di pubblicazione
            )
            response = request.execute()

            # Aggiunta di debug per confermare la risposta API
            if response:
                print(f"Chiamata API eseguita con successo, ottenuti {len(response['items'])} risultati.")
            else:
                print("Errore: La risposta API è vuota o nulla.")
                break

            for item in response['items']:
                video_id = item['id']['videoId']
                video_details = youtube.videos().list(
                    part="snippet,contentDetails,statistics",
                    id=video_id
                ).execute()

                video_data = {
                    'title': item['snippet']['title'],
                    'videoId': video_id,
                    'channelTitle': item['snippet']['channelTitle'],
                    'channelId': item['snippet']['channelId'],
                    'tags': ', '.join(video_details['items'][0]['snippet'].get('tags', [])),
                    'description': item['snippet']['description'],
                    'publishedAt': item['snippet']['publishedAt']  # Aggiunta data di pubblicazione
                }
                videos.append(video_data)

            next_page_token = response.get('nextPageToken')
            print(f"nextPageToken ricevuto: {next_page_token}")  # Debug: verifica il token ricevuto

            if next_page_token:
                save_next_page_token(next_page_token)
            else:
                print(f"Raggiunto il limite di risultati o fine dei risultati: {len(videos)} video trovati.")
                break

            time.sleep(1)

        except googleapiclient.errors.HttpError as e:
            print(f"Errore API HTTP: {e}")
            break

        except Exception as e:
            print(f"Si è verificato un errore generico: {e}")
            break

    print(f"Totale video trovati e processati: {len(videos)}")
    return videos

def save_videos_to_csv(videos, filename="videos.csv"):
    if not videos:
        print("Nessun video da salvare.")
        return
    
    if os.path.exists(filename):
        with open(filename, newline='', encoding='utf-8') as csvfile:
            reader = csv.DictReader(csvfile)
            existing_videos = list(reader)
    else:
        existing_videos = []
    
    existing_video_ids = {video['videoId'] for video in existing_videos}
    new_videos = [video for video in videos if video['videoId'] not in existing_video_ids]

    if new_videos:
        with open(filename, 'w', newline='', encoding='utf-8') as output_file:
            dict_writer = csv.DictWriter(output_file, fieldnames=new_videos[0].keys())
            dict_writer.writeheader()
            dict_writer.writerows(existing_videos + new_videos)
        
        print(f"Salvati {len(new_videos)} nuovi video. Totale video nel file: {len(existing_videos) + len(new_videos)}.")
    else:
        print("Nessun nuovo video da aggiungere.")

# Esempio d'uso
api_key = 'AIzaSyDFwdz2tqp80s6GwZshqvDlPf82_DXvV7o'
tags = ['olympics', 'paralympics', 'paris2024']
max_results = 1000

# Facoltativo: filtro per data di pubblicazione
published_after = "2023-01-01T00:00:00Z"

videos = search_videos_by_tags(api_key, tags, max_results, published_after)

if videos:
    print(f"Trovati {len(videos)} video.")
    save_videos_to_csv(videos)

```

```python
import pandas as pd
import json
import os
from googleapiclient.discovery import build
from langdetect import detect, LangDetectException

# Funzione per autenticarsi all'API di YouTube
def get_youtube_service(api_key):
    return build('youtube', 'v3', developerKey=api_key)

# Funzione per estrarre i commenti e le risposte di un video
def get_video_comments_with_replies(youtube, video_id, max_threads=5):
    comments = []
    try:
        request = youtube.commentThreads().list(
            part='snippet,replies',
            videoId=video_id,
            maxResults=max_threads,
            textFormat='plainText'
        )
        response = request.execute()

        for item in response['items']:
            comment_data = item['snippet']['topLevelComment']['snippet']
            comment = comment_data['textOriginal']
            author = comment_data['authorDisplayName']
            comments.append((comment, author))

            # Aggiungi le risposte al commento principale, se ci sono
            if 'replies' in item:
                for reply in item['replies']['comments']:
                    reply_comment = reply['snippet']['textOriginal']
                    reply_author = reply['snippet']['authorDisplayName']
                    comments.append((reply_comment, reply_author))
    except Exception as e:
        print(f"Errore nell'estrarre i commenti per il video {video_id}: {e}")
    
    return comments

# Funzione per rilevare la lingua di un commento
def detect_language(comment):
    try:
        return detect(comment)
    except LangDetectException:
        return "unknown"

# Funzione per determinare lo sport basato sui tag o descrizione
def extract_sport_from_tags_and_description(tags, description):
    sports_keywords = [
    "archery", "athletics", "badminton", "baseball", "basketball", "beach volleyball",
    "boxing", "canoeing", "canoe slalom", "canoe sprint", "cycling", "bmx", "mountain biking",
    "road cycling", "track cycling", "diving", "equestrian", "fencing", "field hockey",
    "football", "golf", "gymnastics", "artistic gymnastics", "rhythmic gymnastics",
    "trampoline gymnastics", "handball", "judo", "karate", "modern pentathlon", "rowing",
    "rugby sevens", "sailing", "shooting", "skateboarding", "softball", "sport climbing",
    "surfing", "swimming", "synchronized swimming", "table tennis", "taekwondo", "tennis",
    "triathlon", "volleyball", "water polo", "weightlifting", "wrestling", "freestyle wrestling",
    "greco-roman wrestling", "3x3 basketball", "breakdancing", "marathon swimming", "artistic swimming",
    "canoe kayak", "beach handball", "roller speed skating", "surfing", "kiteboarding", "stand up paddling",
    "sport shooting", "pistol shooting", "rifle shooting", "sailing", "open water swimming", "modern pentathlon",
    "equestrian dressage", "equestrian eventing", "equestrian jumping", "sprint canoe", "slalom canoe",
    "track and field", "race walking", "hammer throw", "javelin throw", "discus throw", "pole vault",
    "long jump", "triple jump", "high jump", "steeplechase", "decathlon", "heptathlon", "mixed relay",
    "sabre fencing", "epee fencing", "foil fencing"
]

    
    # Funzione di ricerca sport
    def find_sport(text):
        for sport in sports_keywords:
            if sport.lower() in text.lower():
                return sport
        return None
    
    # Cerca nei tag
    sport_in_tags = find_sport(tags)
    
    # Cerca nella descrizione se non trovato nei tag
    if not sport_in_tags:
        sport_in_description = find_sport(description)
        return sport_in_description if sport_in_description else None
    
    return sport_in_tags

# Funzione per salvare l'avanzamento in un file JSON
def save_progress(progress_file, progress_data):
    with open(progress_file, 'w') as f:
        json.dump(progress_data, f)

# Funzione per caricare il progresso dal file JSON
def load_progress(progress_file):
    if os.path.exists(progress_file):
        with open(progress_file, 'r') as f:
            return json.load(f)
    return {"last_video_index": 0}

# Funzione principale per l'estrazione e analisi
def extract_comments_from_csv(api_key, input_csv, output_csv, progress_file, max_videos=10, max_threads=5):
    # Carica il file CSV
    df = pd.read_csv(input_csv)
    
    # Autenticazione all'API di YouTube
    youtube = get_youtube_service(api_key)
    
    # Carica il progresso
    progress_data = load_progress(progress_file)
    last_video_index = progress_data.get('last_video_index', 0)
    
    # Lista per immagazzinare i dati finali
    results = []
    videos_processed = 0

    # Loop attraverso i video nel CSV a partire dall'ultimo video elaborato
    for index, row in df.iterrows():
        if index < last_video_index:
            continue
        
        # Limita il numero massimo di video da processare
        if videos_processed >= max_videos:
            break
        
        video_id = row['videoId']
        tags = row['tags'] if pd.notna(row['tags']) else ''
        description = row['description'] if pd.notna(row['description']) else ''
        sport = extract_sport_from_tags_and_description(tags, description)
        
        print(f"Estrazione dei commenti per il video {videos_processed + 1}/{max_videos} - ID: {video_id} (Sport: {sport})")
        
        # Estrazione dei commenti e delle risposte
        comments = get_video_comments_with_replies(youtube, video_id, max_threads)

        # Stampa il numero di thread e il numero di commenti totali
        num_comments = len(comments)
        total_length = sum(len(comment) for comment, _ in comments)
        print(f"Numero di commenti estratti: {num_comments}, Lunghezza complessiva: {total_length} caratteri")

        # Analisi della lingua per ciascun commento
        for comment, author in comments:
            lang = detect_language(comment)
            results.append({
                'video_id': video_id,
                'comment': comment,
                'author': author,
                'language': lang,
                'sport': sport
            })
        
        # Salva l'avanzamento corrente nel file JSON
        progress_data['last_video_index'] = index + 1
        save_progress(progress_file, progress_data)

        # Incrementa il contatore dei video elaborati
        videos_processed += 1

    # Salva i risultati in un nuovo file CSV
    results_df = pd.DataFrame(results)
    if os.path.exists(output_csv):
        results_df.to_csv(output_csv, mode='a', header=False, index=False)
    else:
        results_df.to_csv(output_csv, index=False)
    
    # Stampa il numero totale di video processati e la dimensione del file CSV
    print(f"Numero totale di video elaborati: {videos_processed}")
    print(f"Numero di commenti salvati: {len(results_df)}")
    print(f"File CSV salvato: {output_csv}")

# Inserisci qui la tua chiave API di YouTube
api_key = 'AIzaSyDFwdz2tqp80s6GwZshqvDlPf82_DXvV7o'

# Percorso del file CSV di input (videos.csv) e del file di output
input_csv = 'videos.csv'
output_csv = 'video_comments.csv'

# File JSON per tenere traccia dell'avanzamento
progress_file = 'progress.json'

# Esecuzione del programma (modifica il numero di video e di thread per video se necessario)
extract_comments_from_csv(api_key, input_csv, output_csv, progress_file, max_videos=500, max_threads=10)
```

```python
import json
import os

def load_json(filename):
    """Carica il file JSON se esiste, altrimenti restituisce un dizionario vuoto."""
    if os.path.exists(filename):
        with open(filename, 'r', encoding='utf-8') as f:
            return json.load(f)
    else:
        return {}

def save_json(data, filename):
    """Salva il dizionario nel file JSON."""
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)
    print(f"File '{filename}' aggiornato con successo.")

def modify_json(filename, key, value):
    """Modifica il file JSON aggiungendo o aggiornando una chiave con un valore."""
    data = load_json(filename)
    data[key] = value
    save_json(data, filename)

# Esempio di utilizzo
filename = 'next_page_token.json'
key = 'nextPageToken'
value = 'CJADEAA'

modify_json(filename, key, value)
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Caricamento del file CSV
file_path = 'video_comments.csv'  # Inserisci il percorso corretto del file CSV
df = pd.read_csv(file_path)

# Creazione del grafo
G = nx.Graph()
edge_count = 0  # Contatore per il numero di archi

# Creazione degli archi basati su video, sport e lingua
for video_id in df['video_id'].unique():
    # Ottieni tutti gli utenti che hanno commentato lo stesso video
    utenti_video = df[df['video_id'] == video_id]
    
    for i, user1 in utenti_video.iterrows():
        for j, user2 in utenti_video.iterrows():
            if i >= j:  # Evita di ripetere le connessioni
                continue
            
            u1 = user1['author']
            u2 = user2['author']
            sport = user1['sport']
            language = user1['language']
            video = user1['video_id']
            
            # Aggiungi nodo per ogni utente
            G.add_node(u1)
            G.add_node(u2)
            
            # Aggiungi arco tra due utenti se commentano lo stesso video
            if G.has_edge(u1, u2):
                # Se l'arco esiste già, aggiorna gli attributi
                G[u1][u2]['videos'].add(video)
                G[u1][u2]['sports'].add(sport)
                G[u1][u2]['languages'].add(language)
            else:
                # Aggiungi nuovo arco con attributi video, sport, lingua
                G.add_edge(u1, u2, videos={video}, sports={sport}, languages={language})
                edge_count += 1  # Incrementa il contatore degli archi

                # Stampa ogni 5000 archi creati
                if edge_count % 5000 == 0:
                    print(f"Creati {edge_count} archi finora...")
                    print(f"Numero attuale di nodi: {G.number_of_nodes()}")
                    print(f"Numero attuale di archi: {G.number_of_edges()}")

# Fine del ciclo, stampa il totale
print(f"Processo completato con {G.number_of_nodes()} nodi e {G.number_of_edges()} archi.")

# Prima di salvare, convertiamo i set in stringhe negli attributi degli archi
for u, v, data in G.edges(data=True):
    data['videos'] = ', '.join(str(v) for v in data['videos'] if pd.notna(v))  # Converte il set in una stringa e rimuove i NaN
    data['sports'] = ', '.join(str(v) for v in data['sports'] if pd.notna(v))  # Stessa logica per sport
    data['languages'] = ', '.join(str(v) for v in data['languages'] if pd.notna(v))  # Stessa logica per languages

# Salva il grafo in un file GEXF
nx.write_gexf(G, 'grafo_utenti_sport_lingua.gexf')

print("Grafo salvato come 'grafo_utenti_sport_lingua.gexf'.")
```

```python
#CALCOLO GRADO MEDIO E DISTRIBUZIONE GRADO


import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Carica la rete dal file .gexf
G = nx.read_gexf('grafo_utenti_sport_lingua.gexf')

# Calcola il grado di ciascun nodo
degree_sequence = [degree for node, degree in G.degree()]

# Calcola la distribuzione del grado
degree_counts = np.bincount(degree_sequence)
degrees = np.nonzero(degree_counts)[0]
distribution = degree_counts[degrees] / sum(degree_counts)

# Calcola il grado medio
average_degree = np.mean(degree_sequence)
print(f"Grado medio: {average_degree}")

plt.figure(figsize=(10, 6))
plt.loglog(degrees, distribution, marker='o', linestyle='none')
plt.title('Distribuzione del Grado dei Nodi (Log-Log)')
plt.xlabel('Grado')
plt.ylabel('Frazione di Nodi')
plt.grid(True)
plt.show()

```

```python
import networkx as nx

# Carica il file della rete in formato GEXF
G = nx.read_gexf('grafo_utenti_sport_lingua.gexf')

# Trova le componenti connesse della rete
connected_components = list(nx.connected_components(G))

# Seleziona la componente gigante (quella con il maggior numero di nodi)
giant_component = max(connected_components, key=len)

# Crea un subgrafo della componente gigante
G_giant = G.subgraph(giant_component)

# Calcola la lunghezza media dei percorsi minimi per la componente gigante
average_shortest_path_length = nx.average_shortest_path_length(G_giant)
print(f"Lunghezza media dei percorsi minimi nella componente gigante: {average_shortest_path_length}")


# Calcola il coefficiente di clustering medio
average_clustering_coefficient = nx.average_clustering(G)
print(f"Coefficiente di Clustering Medio: {average_clustering_coefficient}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Carica la rete dal file .gexf
G = nx.read_gexf('grafo_utenti_sport_lingua.gexf')

# Numero di nodi e archi nella tua rete
n = G.number_of_nodes()
m = G.number_of_edges()

# Probabilità di collegamento per il modello Erdős-Rényi
p = 2 * m / (n * (n - 1))

# Genera una rete Erdős-Rényi con lo stesso numero di nodi e probabilità di collegamento
G_er = nx.erdos_renyi_graph(n, p)

# Calcola la distribuzione del grado per la tua rete
degree_sequence = [degree for node, degree in G.degree()]
degree_counts = np.bincount(degree_sequence)
degrees = np.nonzero(degree_counts)[0]
distribution = degree_counts[degrees] / sum(degree_counts)

# Calcola la distribuzione del grado per la rete Erdős-Rényi
degree_sequence_er = [degree for node, degree in G_er.degree()]
degree_counts_er = np.bincount(degree_sequence_er)
degrees_er = np.nonzero(degree_counts_er)[0]
distribution_er = degree_counts_er[degrees_er] / sum(degree_counts_er)

# Visualizza il confronto tra le distribuzioni del grado in scala log-log
plt.figure(figsize=(10, 6))
plt.loglog(degrees, distribution, 'bo-', label='Tua Rete')
plt.loglog(degrees_er, distribution_er, 'ro-', label='Erdős-Rényi')
plt.title('Confronto della Distribuzione del Grado (Log-Log)')
plt.xlabel('Grado')
plt.ylabel('Frazione di Nodi')
plt.legend()
plt.grid(True, which="both", ls="--")
plt.show()
```

```python
import networkx as nx
import time

# Funzione per stampare messaggi periodici
def print_progress(message, start_time, interval=600):  # 600 secondi = 10 minuti
    elapsed_time = time.time() - start_time
    if elapsed_time >= interval:
        print(f"Sono ancora vivo e sto facendo questo: {message}")
        return time.time()  # Reset del timer
    return start_time

# Definisci i parametri della rete ER
n = 28967  # Numero di nodi
p = 0.00035  # Probabilità di creazione di un arco (uguale alla densità)

start_time = time.time()
print("Inizio creazione della rete Erdos-Renyi.")

# Crea una rete randomica ER usando il modello Erdos-Renyi
ER_random_graph = nx.erdos_renyi_graph(n, p)

# Controlla il tempo e stampa un messaggio ogni 10 minuti
start_time = print_progress("Sto creando la rete Erdos-Renyi...", start_time)

# Calcola il coefficiente di clustering medio della rete ER
print("Inizio calcolo del coefficiente di clustering medio.")
average_clustering_coefficient_ER = nx.average_clustering(ER_random_graph)
print(f"Coefficiente di Clustering Medio della Rete ER: {average_clustering_coefficient_ER}")

# Controlla il tempo e stampa un messaggio ogni 10 minuti
start_time = print_progress("Sto calcolando il coefficiente di clustering...", start_time)

# Verifica se la rete è connessa
print("Verifico se la rete è connessa.")
if nx.is_connected(ER_random_graph):
    # Calcola la lunghezza media dei cammini minimi
    print("La rete è connessa, inizio calcolo della lunghezza media dei cammini minimi.")
    average_shortest_path_length_ER = nx.average_shortest_path_length(ER_random_graph)
    print(f"Lunghezza Media dei Cammini Minimi della Rete ER: {average_shortest_path_length_ER}")
else:
    print("La rete non è connessa, estraggo la componente gigante.")
    # Estrai la componente gigante
    largest_component = max(nx.connected_components(ER_random_graph), key=len)
    ER_giant_component = ER_random_graph.subgraph(largest_component)

    # Calcola la lunghezza media dei cammini minimi della componente gigante
    print("Inizio calcolo della lunghezza media dei cammini minimi della componente gigante.")
    average_shortest_path_length_ER = nx.average_shortest_path_length(ER_giant_component)
    print(f"Lunghezza Media dei Cammini Minimi della Componente Gigante della Rete ER: {average_shortest_path_length_ER}")

# Controlla il tempo e stampa un messaggio ogni 10 minuti
start_time = print_progress("Sto calcolando la lunghezza media dei cammini minimi...", start_time)

print("Processo completato.")
```

```python
import pandas as pd
from collections import defaultdict
import networkx as nx
from community import community_louvain
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Funzione per caricare il grafo da un file .gexf
def load_graph_from_gexf(file_path):
    G = nx.read_gexf(file_path)
    return G

# Funzione per estrarre gli attributi dagli archi
def extract_edge_attributes(graph):
    edge_attributes = defaultdict(lambda: {'sports': set(), 'languages': set()})
    for u, v, data in graph.edges(data=True):
        sports = data.get('sports', "")
        languages = data.get('languages', "")
        
        # Aggiungi gli sport e le lingue agli attributi degli archi se presenti
        if sports and sports.lower() not in ("", "unknown"):
            for sport in sports.split(", "):
                edge_attributes[(u, v)]['sports'].add(sport)
        if languages and languages.lower() not in ("", "unknown"):
            for language in languages.split(", "):
                edge_attributes[(u, v)]['languages'].add(language)
    
    return edge_attributes

# Carica la rete dal file GEXF
G = load_graph_from_gexf('grafo_utenti_sport_lingua.gexf')

# Rilevamento delle comunità usando l'algoritmo di Louvain
partition = community_louvain.best_partition(G)

# Ordina le comunità per dimensione
community_sizes = defaultdict(int)
for community_id in partition.values():
    community_sizes[community_id] += 1
sorted_communities = sorted(community_sizes.items(), key=lambda x: x[1], reverse=True)

# Identifica le 10 comunità più grandi
top_communities = [community_id for community_id, _ in sorted_communities[:10]]

print("Le 10 comunità più grandi per numero di nodi:")
for community_id, size in sorted_communities[:10]:
    print(f"Comunità {community_id}: {size} nodi")

# Estrai gli attributi degli archi
edge_attributes = extract_edge_attributes(G)

# Analizza gli attributi più presenti nelle 10 comunità più grandi
def analyze_community_attributes(graph, partition, top_communities, edge_attributes):
    attributes_by_community = defaultdict(lambda: {'sports': defaultdict(int), 'languages': defaultdict(int)})  # Struttura: attributes_by_community[community_id][attribute] = count
    for node, community_id in partition.items():
        if community_id in top_communities:
            # Esamina gli archi collegati al nodo
            for neighbor in graph.neighbors(node):
                edge_data = edge_attributes.get((node, neighbor)) or edge_attributes.get((neighbor, node))
                if edge_data:
                    sports = edge_data.get('sports', set())
                    languages = edge_data.get('languages', set())
                    for sport in sports:
                        attributes_by_community[community_id]['sports'][sport] += 1
                    for language in languages:
                        attributes_by_community[community_id]['languages'][language] += 1
    
    for community_id in top_communities:
        print(f"\nComunità {community_id}:")
        for attr_type in ['sports', 'languages']:
            attr_counts = attributes_by_community[community_id][attr_type]
            if attr_counts:
                # Trova l'attributo con il conteggio maggiore
                most_common_attr = max(attr_counts.items(), key=lambda x: x[1])
                print(f"  Attributo {attr_type} più comune: {most_common_attr[0]} (conteggio: {most_common_attr[1]})")

# Esegui l'analisi degli attributi
analyze_community_attributes(G, partition, top_communities, edge_attributes)

# Calcola la sovrapposizione massima per sport e lingua tra due comunità
def analyze_max_overlap(top_communities, graph, partition, edge_attributes, attribute_type):
    for attr in set(a for e in edge_attributes.values() for a in e[attribute_type]):
        
        community_edges = defaultdict(set)  # community_edges[community_id] = set di archi
        for u, v in graph.edges():
            if partition[u] in top_communities and partition[v] in top_communities:
                if edge_attributes.get((u, v)) and attr in edge_attributes[(u, v)][attribute_type]:
                    community_edges[partition[u]].add((u, v))
                    community_edges[partition[v]].add((u, v))

        max_overlap_percentage = 0
        max_overlap_pair = (None, None)
        
        for i, community_id_1 in enumerate(top_communities):
            for community_id_2 in top_communities[i + 1:]:
                edges_in_1 = community_edges.get(community_id_1, set())
                edges_in_2 = community_edges.get(community_id_2, set())
                overlap = len(edges_in_1 & edges_in_2)
                total = len(edges_in_1 | edges_in_2)
                percentage_overlap = (overlap / total) * 100 if total > 0 else 0
                
                # Aggiorna il massimo se necessario
                if percentage_overlap > max_overlap_percentage:
                    max_overlap_percentage = percentage_overlap
                    max_overlap_pair = (community_id_1, community_id_2)

        # Stampa solo se la sovrapposizione è maggiore di 0 e le comunità sono definite
        if max_overlap_percentage > 0 and max_overlap_pair != (None, None):
            print(f"Massima sovrapposizione per {attr}: {max_overlap_percentage:.2f}% tra comunità {max_overlap_pair[0]} e comunità {max_overlap_pair[1]}")

# Sovrapposizione massima per sport
analyze_max_overlap(top_communities, G, partition, edge_attributes, 'sports')

# Sovrapposizione massima per lingua
analyze_max_overlap(top_communities, G, partition, edge_attributes, 'languages')


# Visualizzazione della Rete Focalizzata sulle 10 Comunità
subgraph_nodes = [node for node, community_id in partition.items() if community_id in top_communities]
subgraph = G.subgraph(subgraph_nodes)

pos = nx.spring_layout(subgraph)
fig, ax = plt.subplots(figsize=(12, 10))

# Disegna il sottografo
nx.draw(subgraph, pos, with_labels=False, node_size=50, node_color=[partition[n] for n in subgraph_nodes], cmap=plt.get_cmap('tab10'), edge_color='gray', ax=ax)

# Crea una legenda
handles = []
for community_id in top_communities:
    color = plt.get_cmap('tab10')(community_id % 10)  # Associa un colore a ciascuna comunità
    handle = patches.Patch(color=color, label=f'Comunità {community_id}')
    handles.append(handle)

ax.legend(handles=handles, bbox_to_anchor=(1.05, 1), loc='upper left')
ax.set_title('Rete Focalizzata sulle 10 Comunità')

plt.show()
```

```python
import pandas as pd
import networkx as nx
from collections import defaultdict
import matplotlib.pyplot as plt
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from torch.nn.functional import softmax

# Carica il modello pre-addestrato e il tokenizer
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name)

# Funzione per analizzare il sentimento
def analyze_sentiment(text):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)
    outputs = model(**inputs)
    probs = softmax(outputs.logits, dim=1)
    sentiment_score = torch.argmax(probs).item() + 1  # Ritorna il valore della classe (da 1 a 5)
    return sentiment_score

# Leggi il file CSV e crea un dizionario di commenti per autore
def load_comments_from_csv(csv_file):
    df = pd.read_csv(csv_file)
    comments_by_author = defaultdict(list)
    for _, row in df.iterrows():
        author = row['author']
        comment = row['comment']
        comments_by_author[author].append(comment)
    return comments_by_author

# Raggruppa i commenti per sport usando il grafo e i commenti dai CSV
def group_comments_by_sport_from_graph_and_csv(graph, comments_by_author):
    comments_by_sport = defaultdict(list)
    for u, v, data in graph.edges(data=True):
        sport = data.get('sports', '')
        if sport:  # Escludi sport vuoti
            comments_u = comments_by_author.get(u, [])
            comments_v = comments_by_author.get(v, [])
            comments_by_sport[sport].extend(comments_u)
            comments_by_sport[sport].extend(comments_v)
    return comments_by_sport

# Calcola il sentiment medio per sport
def calculate_average_sentiment_by_sport(comments_by_sport):
    sentiments_by_sport = {}
    for sport, comments in comments_by_sport.items():
        if comments:  # Assicurati che ci siano commenti
            sentiments = [analyze_sentiment(comment) for comment in comments]
            average_sentiment = sum(sentiments) / len(sentiments)
            sentiments_by_sport[sport] = average_sentiment
    return sentiments_by_sport

# Carica la rete dal file .gexf
G = nx.read_gexf('grafo_utenti_sport_lingua.gexf')

# Carica i commenti dal file CSV
comments_by_author = load_comments_from_csv('video_comments.csv')

# Raggruppa i commenti per sport
comments_by_sport = group_comments_by_sport_from_graph_and_csv(G, comments_by_author)

# Calcola il sentiment medio per sport
sentiments_by_sport = calculate_average_sentiment_by_sport(comments_by_sport)

# Filtra gli sport sconosciuti
filtered_sentiments_by_sport = {sport: avg_sentiment for sport, avg_sentiment in sentiments_by_sport.items() if sport}

# Imposta la dimensione della figura
plt.figure(figsize=(10, 8))  # Imposta una dimensione più grande (es. 10x8)

# Visualizza la media del sentiment per sport
sports = sorted(filtered_sentiments_by_sport.keys())
average_sentiments = [filtered_sentiments_by_sport[sport] for sport in sports]

# Crea il grafico a barre orizzontali
plt.barh(sports, average_sentiments, color='skyblue')

# Impostazioni del grafico
plt.title('Sentiment Medio per Sport', fontsize=16)  # Titolo più grande
plt.xlabel('Sentiment Medio', fontsize=14)  # Etichetta asse x più grande
plt.ylabel('Sport', fontsize=14)  # Etichetta asse y più grande
plt.grid(True, axis='x', linestyle='--', alpha=0.7)  # Aggiungi griglia solo sull'asse x per la leggibilità

# Riduci la dimensione del font delle etichette degli sport
plt.yticks(fontsize=10)

# Aggiungi spazio tra le barre e i nomi degli sport
plt.tight_layout()

# Mostra il grafico
plt.show()

# Stampa i risultati
for sport, avg_sentiment in filtered_sentiments_by_sport.items():
    print(f"Sport: {sport}, Sentiment Medio: {avg_sentiment:.2f}")
```

```python
from collections import defaultdict

# Inizializza dizionari per sport e lingue
sport_count = defaultdict(int)
language_count = defaultdict(int)

# Itera attraverso gli archi per contare sport e lingue
for u, v, data in G.edges(data=True):
    sports = data.get('sports', "")
    languages = data.get('languages', "")
    
    # Conta ogni sport (se non è vuoto o 'unknown')
    if sports and sports.lower() not in ("", "unknown"):
        for sport in sports.split(", "):
            sport_count[sport] += 1

    # Conta ogni lingua (se non è vuoto o 'unknown')
    if languages and languages.lower() not in ("", "unknown"):
        for language in languages.split(", "):
            language_count[language] += 1

# Stampa i risultati
print("Conteggio sport:")
for sport, count in sport_count.items():
    print(f"{sport}: {count}")

print("\nConteggio lingue:")
for language, count in language_count.items():
    print(f"{language}: {count}")
```
